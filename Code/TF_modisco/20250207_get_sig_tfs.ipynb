```python
# motifs file
file1 = "count_motifs.csv"
file2 = "profile_motifs.csv"
```

```python
# read in each motif file
motif1 = pd.read_csv(file1)
motif2 = pd.read_csv(file2)
```

```python
sig_tf1 = motif1.loc[motif1["P_Value"]<=0.05, "TF_Name"].tolist()
sig_tf2 = motif2.loc[motif2["P_Value"]<=0.05, "TF_Name"].tolist()
```

```python
# save
with open("sig_tfs_modisco.txt", "w") as f:
    for item in set(sig_tf1 + sig_tf2):
        f.write(item + "\n")
```
